```python
import json
from jsonschema.validators import validator_for

schema_path = "accord_json_schema.json"
with open(schema_path, "r") as fp:
    schema = json.load(fp)

policy_path = "precoa-accord.json"
with open(policy_path, "r") as fp:
    policy = json.load(fp)

# Pick the validator class based on the schema's declared draft.
ValidatorClass = validator_for(schema)
ValidatorClass.check_schema(schema)
validator = ValidatorClass(schema)

errors = sorted(validator.iter_errors(policy), key=lambda e: list(e.absolute_path))

if not errors:
    print("JSON is valid.")
else:
    print(f"Found {len(errors)} validation error(s):")
    for i, error in enumerate(errors, 1):
        path = ".".join(str(p) for p in error.absolute_path) or "<root>"
        print(f"{i}. Path: {path}")
        print(f"   Message: {error.message}")
```
