```python
import pandas as pd
import json
carriage_policies_df = pd.read_csv('carriage-policies-20260914.csv')
agents_df = pd.read_csv('appqore-users-20260824-03.csv')

agents_df['producerID'] = agents_df['producerID'].astype(str).str.strip()
carriage_policies_df['producerID'] = carriage_policies_df['producerID'].astype(str).str.strip()
```

```python
merged_df = pd.merge(carriage_policies_df, agents_df, on='producerID', how='left', suffixes=('_appqore', '_id3'))
```

```python
merged_df = merged_df.rename(
    columns={
        "templateName": "TemplateName",
        "policyData.policyNumber": "PolicyNumber",
        "producerID": "Agent",
        "name": "AgentName",
        "email": "AgentEmail",
    }
)
```

```python
merged_df.to_csv("carriage-policies-users.csv",index=False)
```
